Loading All the library

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
```

Loading the dataset(iris flower dataset)

```python
iris = load_iris()

X = iris.data

feature_names = iris.feature_names
```

Finding the PCA by SVD

```python
X_centered = X - X.mean(axis=0)
U, s, Vt = np.linalg.svd(X_centered)
C1 = Vt.T[:,0]
C2 = Vt.T[:,1]

print(f"The first Principle component is --> {C1}")
print(f"The second principle component is --> {C2}")
```

Projection on the PCA

```python
X_pca = X_centered @ Vt.T[:, :2]
```

Plotting

```python
plt.figure(figsize=(10, 6))
colors = ['red', 'green', 'blue']
for i, species in enumerate(np.unique(iris.target)):
    mask = iris.target == i
    plt.scatter(X_pca[mask, 0], X_pca[mask, 1], 
                color=colors[i], label=iris.target_names[i], alpha=0.7)
plt.xlabel('First Principal Component')
plt.ylabel('Second Principal Component')
plt.title('Iris Dataset: First Two Principal Components')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

Finding PCA using Scikit-learn

```python
# Library load
from sklearn.decomposition import PCA
```

```python
# Create PCA object and fit
pca = PCA(n_components=2) # Reduce to 2 dimensions

X_pca = pca.fit_transform(X) # Learns the principal components from your data and project on those PC

print(X_pca) # projection of the X on PCs
```
